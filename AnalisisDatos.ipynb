```python
import pandas as pd
```

```python
df_energia = pd.read_csv("consumo_energia_iot.csv")
df_energia
```

```python
#* Seleccionamos consumo kwh de la tabla y lo sacamos como objeto pd.serie

consumo_kwh = df_energia["Consumo_kWh"]
serie = pd.Series(consumo_kwh)
serie , type(serie)
```

```python
municipio = df_energia["Municipio"]
estrato = df_energia["Estrato"]
voltaje_promedio = df_energia["Voltaje_Promedio"]
municipio,estrato,voltaje_promedio
```

```python
#Consulta de Lectura Específica (loc): Mediante el indexador .loc[], extrae todos los datos de la fila ubicada en el índice 150.
fila = df_energia.loc[150]
fila
```

```python
#Depuración de Registros Duplicados: Detecta y remueve las filas redundantes con .drop_duplicates() para garantizar que cada registro sea único en la tabla.
df_sin_duplicados = df_energia.drop_duplicates
df_sin_duplicados
```

```python
#Filtrado Booleano Simple: Filtra los registros para identificar predios con un consumo elevado, seleccionando aquellos con Consumo_kWh > 250.0.

df_energia[df_energia["Consumo_kWh"]>250.0]
```

```python
#Filtrado Categórico por Lista (isin): Utiliza .isin() para recuperar únicamente los medidores instalados en los municipios de Chinchiná y Villamaría.
df_energia[df_energia["Municipio"].isin(["Chinchiná", "Villamaría"])]
```

```python
#Filtrado por Intervalo Numérico (between): Utilizando el método .between(), extrae todos los registros cuyo Voltaje_Promedio se mantenga en el rango óptimo de 118.0 a 122.0 voltios inclusive.
df_energia[df_energia["Voltaje_Promedio"].between(118.0, 122.0)]
```

```python
# Consultas Directas con .query(): Aplica una consulta con .query() para filtrar aquellos inmuebles donde Estrato == 'Comercial' y el Estado_Pago == 'Mora'.
df_energia.query("Estrato == 'Comercial' & Estado_Pago == 'Mora'" )
```

```python
#Filtrado Compuesto con Métodos de Texto: Construye una condición lógica con el operador & para encontrar lecturas donde el Tipo_Medidor comience con "B" (.str.startswith()) y las Horas_Pico_kWh sean mayores a 40.0.
filtro = (df_energia["Tipo_Medidor"].str.startswith("B")) & (df_energia["Horas_Pico_kWh"]>40)
df_energia[filtro]
```

```python
#Tratamiento e Imputación de Nulos: Rellena los valores ausentes (NaN) en la columna Consumo_kWh utilizando la media aritmética de dicha variable, e imputa los nulos en Voltaje_Promedio con el valor nominal estándar 120.0 mediante .fillna().
df_fill = df_energia.fillna(
    {
        "Consumo_kWh": df_energia["Consumo_kWh"].mean(),
        "Voltaje_Promedio": 120
    }
)
df_fill
```

```python
#Transformación Condicional con .apply(): Diseña una función de Python que reciba el valor de Consumo_kWh y asigne el nivel tarifario: "Bajo" (menor a 120 kWh), "Medio" (entre 120 y 250 kWh) o "Alto" (mayor a 250 kWh). Utiliza .apply() para crear la columna Nivel_Consumo.
def transformacion(Consumo_kWh):
    if Consumo_kWh < 120.0:
        return "Bajo"
    if Consumo_kWh <= 250.0:
        return "Medio"
    if Consumo_kWh > 250.0:
        return "Alto"

df_energia["Nivel_Consumo"] = df_energia["Consumo_kWh"].apply(transformacion)

df_energia

```

```python
##Renombrado y Reordenamiento de Columnas: Renombra la columna Consumo_kWh a Energia_kWh y Tipo_Medidor a Tecnologia usando .rename(). Reordena las columnas de modo que ID_Medidor y Nivel_Consumo queden al inicio de la tabla.

df_renombrado = df_energia.rename(
    columns ={
        "Consumo_kWh" : "Energia_kWh",
        "Tipo_Medidor" : "Tecnologia"
    }
)
columnas = ["ID_Medidor", "Nivel_Consumo"]
df_ordenado = df_energia[columnas]

df_renombrado,df_ordenado
```

```python
#Agregación por Municipio: Agrupa los datos por Municipio y calcula simultáneamente el consumo promedio de Energia_kWh y el total acumulado de Horas_Pico_kWh mediante el método .agg().
grouped = df_renombrado.groupby("Municipio")
grouped.groups
aggregate = grouped.agg(
    {
        "Energia_kWh" : "mean",
        "Horas_Pico_kWh": "sum"
    }
)
aggregate
```

```python
#Métricas de Rango Personalizadas: Diseña una función personalizada rango (series.max() - series.min()) y aplícala con .agg() para calcular la dispersión del consumo de energía y del voltaje promedio agrupando por Estrato.


def rango(serie):
    return serie.max() - serie.min()

aggregate_data_custom = grouped.agg(
    {
        "Voltaje_Promedio" : rango
    }
)

aggregate_data_custom
```

```python
#Agrupamiento Multinivel: Realiza una agrupación simultánea por Municipio y Estrato, obteniendo un resumen que presente el promedio de Energia_kWh y el promedio de Voltaje_Promedio para cada combinación de sectores.
grouped_multi = df_renombrado.groupby(["Municipio", "Estrato"])
promedio = grouped_multi.agg(
    {
        "Energia_kWh" : "mean",
        "Voltaje_Promedio" : "mean"
    }
)
grouped_multi.groups,promedio
```
